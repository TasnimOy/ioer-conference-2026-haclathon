### Digitalisation meets Transformative Governance
**A collaborative data challenge at the 4th IOER Conference "Space & Transformation"**

Welcome to the digital workspace of the **Sustainability Transformation HaCLAthon** (Format F5 of the [4th IOER Conference 2026](https://conference.ioer.info)). 

Unlike traditional hackathons that require exhausting overnight coding marathons, the HaCLAthon format is designed as a collaborative, long-term, asynchronous data challenge. 

> Ha**CLA**thon = **C**ollaborative, **L**ong-term, **A**synchronous Hackathon

We bring together spatial data analysts, planners, coders, and domain experts to explore how Artificial Intelligence and Open Data can foster participation and sustainability.

This Jupyter Book serves as an interactive counterpart to traditional conference proceedings. All accepted contributions will be merged into this living publication, and all contributors will become co-authors of the final, citable data artifact.

### How to participate

The collaboration period is open now and runs until the conference begins. Anyone who wishes to participate is invited, and formal conference registration is not a requirement to contribute to the book.

```{figure} resources/timeline-hackathon.svg
:name: hackathon-timeline

We welcome contributions ranging from technical code to conceptual data stories. You do not need to be a programming expert. We provide access to curated IOER datasets, ready-to-use cloud environments, and a visual text editor to accommodate different technical backgrounds.

````

The general workflow follows these milestones:

1. **Collaborate:** Work on your idea asynchronously using our cloud environments, Git, or simple file uploads.
2. **Publish:** We review and merge submissions until 10 September 2026.
3. **Present:** Join us in Dresden on 23-24 September 2026 to discuss the results (conference attendance is optional).
4. **Share:** The book will be archived and made available with a DOI after 24 October 2026.

### The Challenge Topics
We invite contributions that explore the intersection of **Digitalisation** and **Transformative Governance**. Specifically, we are looking for data-driven insights, workflows, or visualizations addressing:

*   **Urban Climate Resilience:** Heat adaptation, flood risk, and green infrastructure.
*   **Circularity:** Material flows, recycling potentials, and circular construction.
*   **Land Use Change:** Monitoring settlement dynamics and sprawl.
*   **Material Cadastres:** Mining the anthroposphere for future resources.


### Why Participate?
*   **Create a Lasting Impact:** Your code and analysis will not disappear after the conference. It becomes part of a citable Open Science product.
*   **Low Barrier to Entry:** You do not need to be a Python expert. We provide a "Starter Kit" featuring the IOER Monitor API, allowing you to pull and visualize spatial data immediately.
*   **Interdisciplinary:** We welcome contributions ranging from technical code (Python/R) to conceptual data stories and visual mapping.
*   **Present at the Conference:** Selected teams will be invited to pitch their results during a dedicated session at the main conference in Dresden (Sept 2026).

### About the Infrastructure
This project is powered by the IOER Research Data Centre (IOER-FDZ). We advocate for the <abbr title="Findable, Accessible, Interoperable, Reusable">FAIR</abbr> principles. Computational reproducibility is ensured by running environments via [Jupyter4NFDI](https://base4nfdi.de/projects/jupyter4nfdi). Technical architecture details can be found in the [Developers](../DEVELOPERS) section.

 ```{admonition} Recommended citation
:class: attention
**TBD (2026).** _Sustainability Transformation HaCLAthon_. IOER Conference 2026, JupyterBook Code and Data Materials, [URL](https://hack.conference.ioer.info/), [DOI](https://doi.org/TBD).
```

**Publication Date**: September 2026 (announced)

<div class="fig-no-shadow footer-logos" style="text-align: center; display: flex; justify-content: center; align-items: center; gap: 20px; flex-wrap: wrap; margin-top: 20px;">
<a href="https://fdz.ioer.info" target="_blank" rel="noopener noreferrer">
<img class="light-img logo-img" src="../_static/images/FDZ-Logo_EN_RGB-clr_bg-sol_mgn-full_h200px_web.svg" alt="IOER FDZ Logo" style="height: 100px; width: auto;">
<img class="dark-img logo-img" src="../_static/images/FDZ-Logo_EN_RGB-wht_bg-tra_mgn-full_h200px_web.svg" alt="IOER FDZ Logo" style="height: 100px; width: auto;">
</a>
<a href="https://dlgs.ioer.info" target="_blank" rel="noopener noreferrer">
<img class="light-img logo-img" src="../_static/images/DLGS_logo_clr_mgn-full_h200px.svg" alt="DLGS Logo" style="height: 100px; width: auto;">
<img class="dark-img logo-img" src="../_static/images/DLGS_logo_clr_mgn-full_h200px-mono.svg" alt="DLGS Logo" style="height: 100px; width: auto;">
</a>
</div>
